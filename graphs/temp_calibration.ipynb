```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
thermistor = [14.6,
14.3,
14.05,
13.8,
13.5,
13,
12.8,
12.5,
12.25,
12,
11.9,
11.5,
11.1,
10.7,
10.1]

truth = [20.1,
19.8,
19.5,
19.3,
19.1,
18.8,
18.5,
18.3,
18.1,
17.9,
17.6,
17.3,
17,
16.7,
16.3]
```

```python
thermistor = np.array(thermistor)

coefficients = np.polyfit(thermistor, truth, 1)
slope, intercept = coefficients

best_fit_line = slope * thermistor + intercept

plt.scatter(thermistor, truth, color='black')
plt.plot(thermistor, best_fit_line, color='#9191F3', label=f'Best Fit Line\nslope: {slope:.3f}, intercept: {intercept:.3f}')
plt.xlabel("Thermistor Reading [C]") 
plt.ylabel("Actual Temperature [C]")
plt.grid()
plt.legend()
plt.show()
#plt.savefig('Temp_calibration.png', transparent=True)
```
