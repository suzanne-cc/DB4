```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
volume =  np.array([0, 32, 77, 114, 148, 181, 214, 243, 270])
flow_rate =  np.array([0, 3.2, 7.7, 11.4, 14.8, 18.1, 21.4, 24.3, 27])

coefficients = np.polyfit(flow_rate, volume, 1)
slope, intercept = coefficients

best_fit_line = slope * flow_rate + intercept

plt.scatter(flow_rate, volume, color='black')
plt.plot(flow_rate, best_fit_line, color='#9191F3', label=f'Best Fit Line\nslope: {slope:.3f}, intercept: {intercept:.3f}')
plt.xlabel("Flow Rate [ml/s]") 
plt.ylabel("Volume [ml]")
plt.grid()
plt.legend()
#plt.show()
plt.savefig('OD_pump_calibration.png', transparent=True)
```
